```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.feature_selection import SelectKBest, chi2

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

RANDOM_STATE = 42

sns.set_theme(style="whitegrid")
```

```python
df = data.copy()

print("Dataset information:")
df.info()

print("\nDescriptive statistics:")
display(df.describe(include="all"))

print("\nTarget distribution:")
display(df["label"].value_counts())

print("\nTarget distribution (%):")
display((df["label"].value_counts(normalize=True) * 100).round(2))
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

df = df.drop_duplicates().reset_index(drop=True)

print("Dataset shape after removing duplicates:", df.shape)
```

```python
# Convert text labels into numerical values

df["label"] = df["label"].map({
    "ham": 0,
    "spam": 1
})

print("Labels after encoding:")
print(df["label"].value_counts())

display(df.head())
```

```python
# Independent variable (X)
X = df["message"]

# Dependent variable (y)
y = df["label"]

print("Independent variable (X):", X.shape)
print("Dependent variable (y):", y.shape)
```

```python
# Visualization 1: Count of Ham and Spam Messages

plt.figure(figsize=(6, 4))

sns.countplot(data=df, x="label")

plt.title("Distribution of Ham and Spam Messages")
plt.xlabel("Message Label (0 = Ham, 1 = Spam)")
plt.ylabel("Number of Messages")

plt.show()
```

```python
# Create a new column for message length

df["message_length"] = df["message"].apply(len)

print("Message length details:")
display(df[["label", "message", "message_length"]].head())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="message_length",
    hue="label",
    bins=30,
    kde=True
)

plt.title("Message Length Distribution")
plt.xlabel("Message Length")
plt.ylabel("Number of Messages")

plt.show()
```

```python
# Split the dataset into training and testing data

X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
# Convert text messages into numerical values using TF-IDF

tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=5000
)

X_train_tfidf = tfidf.fit_transform(X_train)

X_test_tfidf = tfidf.transform(X_test)

print("Training data shape:", X_train_tfidf.shape)
print("Testing data shape:", X_test_tfidf.shape)
```

```python
# Feature Selection using Chi-Square

selector = SelectKBest(score_func=chi2, k=3000)

X_train_selected = selector.fit_transform(X_train_tfidf, y_train)

X_test_selected = selector.transform(X_test_tfidf)

print("Training data after feature selection:", X_train_selected.shape)
print("Testing data after feature selection:", X_test_selected.shape)
```

```python
# Initialize Machine Learning Models

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(random_state=RANDOM_STATE),
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=RANDOM_STATE
    )
}

print("Machine Learning Models initialized successfully!")
```

```python
# Train all Machine Learning Models

for name, model in models.items():
    model.fit(X_train_selected, y_train)
    print(name, "training completed successfully!")
```

```python
# Predict using all trained models

predictions = {}

for name, model in models.items():
    y_pred = model.predict(X_test_selected)
    predictions[name] = y_pred

    print(name, "prediction completed successfully!")
```

```python
# Evaluate Machine Learning Models

for name, y_pred in predictions.items():

    print("\nModel:", name)

    print("Accuracy:", accuracy_score(y_test, y_pred))

    print("Precision:", precision_score(y_test, y_pred))

    print("Recall:", recall_score(y_test, y_pred))

    print("F1 Score:", f1_score(y_test, y_pred))
```

```python
# Confusion Matrix for all models

for name, y_pred in predictions.items():

    cm = confusion_matrix(y_test, y_pred)

    plt.figure(figsize=(5, 4))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["Ham", "Spam"],
        yticklabels=["Ham", "Spam"]
    )

    plt.title("Confusion Matrix - " + name)
    plt.xlabel("Predicted Label")
    plt.ylabel("Actual Label")

    plt.show()
```

```python
# Compare all Machine Learning Models

results = []

for name, y_pred in predictions.items():

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1 Score": f1_score(y_test, y_pred)
    })

results_df = pd.DataFrame(results)

display(results_df.round(4))
```

```python
# Find the best model based on accuracy

best_model = results_df.loc[
    results_df["Accuracy"].idxmax()
]

print("Best Performing Model:")
print(best_model)
```
